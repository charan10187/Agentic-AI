# Day 2 — HTTP

    1. What is HTTP?
    HTTP (HyperText Transfer Protocol) is a protocol that defines how a client and server communicate through requests and responses.
    Basic flow:
        Client
        ↓
        HTTP Request
        ↓
        Server
        ↓
        HTTP Response
        ↓
        Client

    2. Client and Server
        Client
        The application that makes the request.
        Examples:
        Browser
        Python program
        Mobile app
        Frontend

    3. HTTP Request
        An HTTP request can contain:
        Request
        ├── Method
        ├── URL
        ├── Headers
        └── Body

    4. HTTP Methods
        You learned the basic methods:
        GET
        POST
        PUT
        PATCH   
        DELETE

        Your current mental model:
        GET     → Get data
        POST    → Send/create data
        PUT     → Replace/update
        PATCH   → Partially update
        DELETE  → Delete

    5. POST
        You specifically clarified POST.
        Example:
        POST /users

        Body:
        {
            "name": "Charan",
            "age": 22
        }

        Meaning:
        Send this data to the /users endpoint, commonly to create a new user.

        You now understand that:
        POST

        is the method,
        /users

        is the endpoint/path,
        and:
        {
            "name": "Charan",
            "age": 22   
        }

        is the request body.

    6. Headers
        Headers contain additional information about the request or response.
        Example:
        Content-Type: application/json

        This tells the server:
        The request body contains JSON.

    7. Request Body
    The body contains the actual data being sent.
    For example:

        {
            "name": "Charan",
            "age": 22
        }

        Important distinction:
        Request body  → Client → Server
        Response body → Server → Client

    8. HTTP Response
        A response can contain:
        Response
        ├── Status Code
        ├── Headers
        └── Body

    9. HTTP Status Codes
    You learned the five major categories:
    1xx → Informational
    2xx → Success
    3xx → Redirection
    4xx → Client Error
    5xx → Server Error

    2xx — Success
    The server successfully received, understood, and processed the client's request.

    Important:
    200 → OK
    201 → Created

    3xx — Redirection
    The server is telling the client that the requested resource is associated with another location or that further action is needed.

    Important:
    301 → Moved Permanently

    4xx — Client Error
    The server received the client's request, but there is a problem with the request or the client's access to the requested resource.

    Important:
    400 → Bad Request
    401 → Unauthorized
    403 → Forbidden
    404 → Not Found
    429 → Too Many Requests

    5xx — Server Error
    The server encountered an error while trying to process a valid client request.

    Important:
    500 → Internal Server Error
    503 → Service Unavailable

    The most important thing you've learned
    You can now understand this:
                        HTTP REQUEST
                            │
                            ▼
    Client ──────────────────────────► Server
            POST /users
            Header
            JSON Body
                            │
                            ▼
                        HTTP RESPONSE
                            │
    Client ◄────────────────────────── Server
            201 Created
            Header
            JSON Body


    JSON       → How data is represented
    HTTP       → How client/server communicate
    Method     → What action is requested
    Endpoint   → Which resource/API is being accessed
    Headers    → Additional information
    Body       → Actual data
    Status     → Result of the request


# Day - APIs


    1. What is an API?
    API = Application Programming Interface
    A simple definition:
    An API is an interface that allows one software application to communicate with another software application according to a defined set of rules.

    Don't think of an API as a programming language or a server.
    Think of it as a communication interface.
    Real-world analogy
    Imagine a restaurant:
    You → Waiter → Kitchen

    You don't walk into the kitchen and directly manipulate the food.
    You tell the waiter:
    "I want biryani."

    The waiter takes your request to the kitchen and brings the result back.
    Similarly:
    Your Python program → API → Server
                            ↓
    Your Python program ← API ← Server

    The API is the interface between your application and the service.
    2. Why do we need APIs?
    Suppose you are building a weather application.
    You don't want to build your own weather-monitoring infrastructure.
    Instead, another company might provide weather data through an API.
    Your application can ask:
    GET /weather

    The service might return:
    {
        "city": "Hyderabad",
        "temperature": 28,
        "humidity": 72
    }

    Your application can then display:
    Hyderabad
    28°C
    72% humidity

    So:
    Your Application
        ↓
        API
        ↓
    Weather Service
        ↓
        JSON
        ↓
    Your Application

    3. API is NOT the same as HTTP
    This distinction is important.
    HTTP
    HTTP defines how communication happens.
    Request → Response

    API
    The API defines what your application can ask the service to do and how it should ask.
    For example:
    GET /users
    GET /users/10
    POST /users
    DELETE /users/10

    These are API endpoints.
    So:
    HTTP
    ↓
    Communication rules

    API
    ↓
    Available operations/interface

    Most modern web APIs use HTTP.
    4. What is an endpoint?
    An endpoint is a specific URL/path through which you access a particular API resource or operation.
    For example:
    https://example.com/users

    Here:
    https://example.com

    is the server/base URL.
    And:
    /users

    is the endpoint path.
    Another endpoint might be:
    /users/101

    which could represent a specific user.
    So you might have:
    GET /users

    → Get all users
    and:
    GET /users/101

    → Get user 101
    5. API request
    An API request can look conceptually like this:
    GET /users

    The server receives it and returns:
    200 OK

    with:
    [
        {
            "id": 1,
            "name": "Charan"
        },
        {
            "id": 2,
            "name": "Alex"
        }
    ]

    ↓

    ↓
